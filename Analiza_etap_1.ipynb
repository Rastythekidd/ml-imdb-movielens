Notebook uruchamia ten sam kod co skrypty projektu. Komórki nie zostały zapisane z wykonanymi wyjściami; gotowe wyniki są w results i PDF. Nie trenujemy modeli. Zależności: requirements.txt. Notebook wymaga dodatkowo środowiska Jupyter lub VS Code z rozszerzeniem Jupyter.


```python
from pathlib import Path
import contextlib, io, json, runpy
from IPython.display import display, Image, Markdown
root = Path.cwd()
assert (root / 'src/analysis.py').exists(), 'Otwórz notebook w katalogu projektu.'
```

## 1. Kontrola jakości i obliczenia

Niezgodność hashy lub wartości zatrzyma wykonanie.


```python
log = io.StringIO()
with contextlib.redirect_stdout(log):
    runpy.run_path(str(root / 'src/analysis.py'), run_name='__main__')
summary = json.loads((root / 'results/summary.json').read_text(encoding='utf-8'))
print(json.dumps(summary['quality'], ensure_ascii=False, indent=2))
```

## 2. MovieLens

Jedna obserwacja to ocena użytkownik-film. Pliki MovieLens zawierają te same 200 ocen.


```python
print(json.dumps(summary['movielens'], ensure_ascii=False, indent=2))
```

## 3. IMDb

Równowaga 100/100 wynika z doboru próby. Metadane obejmują także tytuły inne niż typ movie.


```python
print(json.dumps(summary['imdb'], ensure_ascii=False, indent=2))
```

## 4. Wykresy


```python
runpy.run_path(str(root / 'src/render_figures.py'), run_name='__main__')
for figure in sorted((root / 'results/figures').glob('*.png')):
    display(Markdown('### ' + figure.stem))
    display(Image(filename=str(figure), width=780))
```
